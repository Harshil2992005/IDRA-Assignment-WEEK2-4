## Observations & Key Findings

**1. Dataset Overview**
The dataset has 350 records and 22 columns. Only Promotion column was having missing values (184 rows out of 350).
I filled them with "No Promotion" because on those days simply no offer was running. There were no duplicate records in the dataset.

**2. Revenue Distribution**
Average revenue is around ₹86,831 but it ranges from ₹20,540 to ₹1,94,018.
Skewness of revenue is +0.86 which means most of the branch-days earn normal revenue and only few days earn very high revenue.

**3. Some Days Are in Loss**
Profit is positive for most of the records but the minimum profit is -₹6,924.
This means some operating days ended in loss, so not every day is profitable for the branches.

**4. Premium Stores Are The Best Performers**
Premium store type has average revenue of ₹152,867 which is almost 3 times of Express stores (₹52,327).
Their average profit (₹62,890) is nearly 5 times of Express profit (₹12,742). Clearly Premium format is winning.

**5. Region and Branch Comparison**
South region is performing slightly better with average revenue ₹88,320 and West region is a little behind with ₹84,127,
but overall all three regions are very close to each other.
Bengaluru branch has the highest average revenue (₹90,732) and Mumbai is at the bottom (₹80,779) with lowest profit also.
In customer ratings, Pune is the best branch (4.54) and Kochi is the lowest (4.42).

**6. What Drives Revenue**
Correlation matrix shows Revenue is strongly connected with Food_Cost (0.98), Profit (0.97), Operating_Cost (0.96),
Customers (0.88) and Orders (0.86). So basically when more customers come and more orders happen, revenue increases automatically.

**7. Marketing Alone Is Not Enough**
Marketing Spend is showing only moderate correlation with Revenue (0.43) and Profit (0.35).
So spending more money on marketing does not guarantee high revenue.

**8. Service Quality and Weather Effect**
Delivery time and customer rating are having a negative relation (-0.44), which means slow delivery reduces the rating.
Customer rating is having almost zero connection with revenue (0.008), and weather columns (temperature, humidity)
are not affecting any business metric. Revenue depends more on footfall than weather or service scores.

## 1. Import the module and create an empty list

`csv` is a normal Python module that helps us save and load our data from the file.

We store all expenses inside one list. Each expense is a small list:

`[date, category, amount, note]`


```python
import csv

expenses = []

print("csv module imported successfully.")
print("Empty list is ready to store expenses:", expenses)
```

## 2. Load and save functions (file handling)

- `load()` reads the CSV file at the start so old records come back.
- `save()` writes all expense records back to the CSV file after every change.

If the file does not exist yet, the `try/except FileNotFoundError` catches it and we simply start with an empty list - no crash.


```python
def load():
    try:
        expenses.clear()
        with open("expenses.csv") as f:
            for row in csv.reader(f):
                if len(row) == 4 and row[0] != "Date":
                    expenses.append(row)
        print("Loaded", len(expenses), "expense record(s) from expenses.csv")
    except FileNotFoundError:
        print("No file found - starting a fresh list")


def save():
    with open("expenses.csv", "w", newline="") as f:
        w = csv.writer(f)
        w.writerow(["Date", "Category", "Amount", "Note"])
        for e in expenses:
            w.writerow(e)
    print("Saved all", len(expenses), "expense(s) to expenses.csv")


print("load() and save() functions are ready.")
```

## 3. Function to add a new expense

We ask the user for:

1. date (dd/mm/yyyy)
2. category (food, travel, rent, etc.)
3. amount (a number)
4. note (optional)

The amount is inside `try/except ValueError` - if the user types letters, we show a friendly message instead of crashing. A negative amount is also rejected. After adding, `save()` writes the data to the file right away.


```python
def add():
    date = input("Enter date (dd/mm/yyyy): ")
    cat = input("Enter category (food, travel, rent, etc.): ")

    try:
        amt = float(input("Enter amount: "))
        if amt < 0:
            print("Amount cannot be negative. Please try again.")
            return
    except ValueError:
        print("Invalid amount. Please type a number.")
        return

    note = input("Enter note (optional): ")

    expenses.append([date, cat, amt, note])
    save()
    print("Expense added:", date, "|", cat, "|", amt, "|", note)


print("add() function is ready.")
```

## 4. Function to view all expenses with the total

Shows every expense line by line and adds up all the amounts to print the **total amount spent**.


```python
def view():
    if not expenses:
        print("No expenses yet. Add one first.")
        return

    total = 0
    print("Date | Category | Amount | Note")
    print()
    for e in expenses:
        print(e[0], "|", e[1], "|", e[2], "|", e[3])
        total = total + float(e[2])

    print("-------------------------------------")
    print("Total amount spent:", total)


print("view() function is ready.")
```

## 5. Function for the category-wise summary

Uses a dictionary to group all expenses by category and prints the total spent in every category.


```python
def summary():
    if not expenses:
        print("No expenses yet. Add one first.")
        return

    d = {}
    for e in expenses:
        cat = e[1]
        if cat not in d:
            d[cat] = 0
        d[cat] = d[cat] + float(e[2])

    print("Category-wise summary:")
    for cat in d:
        print(cat, "->", d[cat])


print("summary() function is ready.")
```

## 6. Test drive - see all features working with printed output

In Jupyter it is not comfortable to type answers inside a cell, so here we run the functions with a few **sample expenses**. This proves every assignment requirement works:

1. add a new expense (shown here as sample data)
2. view all expenses with the total amount spent
3. category-wise summary

Running this cell creates `expenses.csv` inside this folder.


```python
# sample expenses added the same way the add() function works
sample = [
    ["14/01/2026", "food",  200, "samosa"],
    ["15/01/2026", "travel", 120, "auto"],
    ["16/01/2026", "food",  350, "dinner"],
]

for s in sample:
    expenses.append(s)

print("Added", len(sample), "sample expenses.")
save()
print()

print("1. View all expenses with the total amount spent")
view()
print()

print("2. Category-wise spending summary")
summary()
```

## 7. The real main menu program (the actual assignment)

This is the **menu driven** part of the assignment. It shows a menu again and again until the user chooses **4 Exit**. Invalid choices are handled without crashing.

> Tip: inside Jupyter this cell waits for typing (`input()`), so it is best to run the whole program as a normal Python file in the terminal - `python expense_tracker.py`.


```python
load()

while True:
    print()
    print("1. Add expense   2. View expenses   3. Summary   4. Exit")
    ch = input("Enter your choice: ")

    if ch == "1":
        add()
    elif ch == "2":
        view()
    elif ch == "3":
        summary()
    elif ch == "4":
        print("Goodbye! Thanks for using the Expense Tracker.")
        break
    else:
        print("Invalid choice. Please pick 1, 2, 3 or 4.")
```

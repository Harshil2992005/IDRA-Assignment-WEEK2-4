## 1. Dataset Inspection
load data set

```python
import pandas as pd
import numpy as np

data = pd.read_csv("Day16_Student_Wellbeing_Survey.csv")

print("Shape -> Rows :", data.shape[0], "| Columns :", data.shape[1])
print()
print("First 3 rows:")
print(data.head(3).to_string())
print()
print("Column Data Types:")
print(data.dtypes.to_string())
```

## 2. Mean, Median, Mode

**Mean** = sum / count , **Median** = middle value , **Mode** = most frequent value

```python
cols = ["Weekly_Study_Hours",
        "Average_Sleep_Hours",
        "Daily_Screen_Time_Hours",
        "Stress_Score",
        "Academic_Readiness_Score"]

for col in cols:
    print(col)
    print("Mean--> Average value:", round(data[col].mean(), 2))
    print("Median--> Middle value:", round(data[col].median(), 2))
    print("Mode--> most frequent value:", data[col].mode()[0])
    print()
```

## 3. Dispersion - Range, Variance, Std Dev, Q1, Q3, IQR

**Range** = max - min , **Variance** = spread from mean , **Std Dev** = sqrt(variance) , **IQR** = Q3 - Q1

```python
for col in cols:
    q1 = data[col].quantile(0.25)
    q3 = data[col].quantile(0.75)
    iqr = q3 - q1

    print(col)
    print("Range--> ", round(data[col].max() - data[col].min(), 2))
    print("Variance--> ", round(data[col].var(), 2))
    print("Std Dev--> ", round(data[col].std(), 2))
    print("Q1--> ", round(q1, 2), "| Q3--> ", round(q3, 2), "| IQR--> ", round(iqr, 2))
    print()

most_sd = max(round(data[c].std(), 2) for c in cols)
for c in cols:
    if round(data[c].std(), 2) == most_sd:
        print("Variable with greatest variability (highest Std Dev):", c, "| SD =", most_sd)
```

## 4. Outlier Detection using IQR Method

**Lower Fence** = Q1 - 1.5*IQR , **Upper Fence** = Q3 + 1.5*IQR . Any value outside is an outlier.

```python
outlier_cols = ["Weekly_Study_Hours",
                "Daily_Screen_Time_Hours",
                "Commute_Time_Minutes",
                "Monthly_Discretionary_Spending"]

for col in outlier_cols:
    q1 = data[col].quantile(0.25)
    q3 = data[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = data[(data[col] < lower) | (data[col] > upper)]

    print(col)
    print("Lower fence:", round(lower, 2))
    print("Upper fence:", round(upper, 2))
    print("Outliers:", len(outliers))
    print()
```

## 5. Mean vs Median - Before and After Removing Outliers

Removing outliers from Monthly_Discretionary_Spending to see the effect on mean vs median.

```python
col = "Monthly_Discretionary_Spending"

q1 = data[col].quantile(0.25)
q3 = data[col].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

clean = data[(data[col] >= lower) & (data[col] <= upper)]

print("Rows removed:", len(data) - len(clean))
print()
print("Before -> Mean:", round(data[col].mean(), 2), "| Median:", round(data[col].median(), 2))
print("After  -> Mean:", round(clean[col].mean(), 2), "| Median:", round(clean[col].median(), 2))
print()
print("Mean changed a lot,median is more robust or stable")
```

## 6. Probability

Event A = Part-time job (Yes) , B = Stress >= 7 , C = Scholarship (Yes) , D = Exercise >= 3 days

**P(Event)** = favourable / total

```python
A = data["Part_Time_Job"] == "Yes"
B = data["Stress_Score"] >= 7
C = data["Scholarship"] == "Yes"
D = data["Exercise_Days_Per_Week"] >= 3
total = len(data)

Prob_A = A.sum() / total
Prob_B = B.sum() / total
Prob_C = C.sum() / total
Prob_D = D.sum() / total

print("Students -> A:", A.sum(), "| B:", B.sum(), "| C:", C.sum(), "| D:", D.sum())
print("P(A) =", round(Prob_A, 4))
print("P(B) =", round(Prob_B, 4))
print("P(C) =", round(Prob_C, 4))
print("P(D) =", round(Prob_D, 4))
```

```python
Prob_A_and_B = (A & B).sum() / total
Prob_A_or_B = (A | B).sum() / total
Prob_A_given_B = Prob_A_and_B / Prob_B
Prob_B_given_A = Prob_A_and_B / Prob_A

print("P(A and B) =", round(Prob_A_and_B, 4))
print("P(A or B) =", round(Prob_A_or_B, 4))
print("P(A|B) =", round(Prob_A_given_B, 4))
print("P(B|A) =", round(Prob_B_given_A, 4))
```

```python
year1 = data["Year_of_Study"] == 1
year4 = data["Year_of_Study"] == 4
both = (year1 & year4).sum()

print("Students in Year 1 AND Year 4 together:", both)
print("P(Year 1 and Year 4) =", round(both / total, 4))
print("Joint is 0 -> MUTUALLY EXCLUSIVE (can't be in both years)")
print()

product = Prob_A * Prob_B
print("Independence check:")
print("P(A and B)  =", round(Prob_A_and_B, 4))
print("P(A) x P(B) =", round(product, 4))
print("Not equal -> events A and B are DEPENDENT")
```

## 7. Bayes' Theorem Verification

P(A|B) = [ P(B|A) x P(A) ] / [ P(B|A) x P(A) + P(B|not A) x P(not A) ]

```python
not_A = ~A
P_not_A = not_A.sum() / total
P_B_given_notA = (B & not_A).sum() / not_A.sum()

numerator = Prob_B_given_A * Prob_A
denominator = numerator + (P_B_given_notA * P_not_A)
bayes = numerator / denominator

print("P(B|A) =", round(Prob_B_given_A, 4))
print("P(B|not A) =", round(P_B_given_notA, 4))
print("P(not A) =", round(P_not_A, 4))
print()
print("Bayes result  P(A|B):", round(bayes, 4))
print("Direct result P(A|B):", round(Prob_A_given_B, 4))
print("Both match -> Bayes' theorem VERIFIED")
```

## 8. Normal Distribution & Z-Scores

**Z** = (X - Mean) / Std Dev . Rule: ~68% within 1 SD, 95% within 2 SD, 99.7% within 3 SD.

```python
score = data["Academic_Readiness_Score"]
mu = score.mean()
sigma = score.std()

top = data.loc[score.idxmax()]
low = data.loc[score.idxmin()]

z_top = (top["Academic_Readiness_Score"] - mu) / sigma
z_low = (low["Academic_Readiness_Score"] - mu) / sigma

print("Mean =", round(mu, 2), "| Std Dev =", round(sigma, 2))
print("Top student:", top["Student_ID"], "score", top["Academic_Readiness_Score"], "Z =", round(z_top, 2))
print("Low student:", low["Student_ID"], "score", low["Academic_Readiness_Score"], "Z =", round(z_low, 2))
print()

w1 = ((score >= mu - sigma) & (score <= mu + sigma)).mean() * 100
w2 = ((score >= mu - 2*sigma) & (score <= mu + 2*sigma)).mean() * 100
w3 = ((score >= mu - 3*sigma) & (score <= mu + 3*sigma)).mean() * 100

print("Within 1 SD -> Expected 68% | Actual", round(w1, 2), "%")
print("Within 2 SD -> Expected 95% | Actual", round(w2, 2), "%")
print("Within 3 SD -> Expected 99.7% | Actual", round(w3, 2), "%")
print("Very close -> score is approximately NORMAL")
print()
print("Z-score interpretation:")
print("Top student Z =", round(z_top, 2), "-> about", str(round(z_top, 2)), "standard deviations ABOVE average (very good performance)")
print("Low student Z =", round(z_low, 2), "-> about", str(abs(round(z_low, 2))), "standard deviations BELOW average (extreme low, very rare)")
```

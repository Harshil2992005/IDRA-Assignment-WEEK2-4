```python
import pandas as pd
```

```python
orders = pd.read_csv("Day9_Orders.csv")
orders.head()
```

```python
customers = pd.read_csv("Day9_Customers.csv")
#customers.info()
customers.head()
```

```python
products = pd.read_csv("Day9_Products.csv")
#products.info()
products.head() 
```

```python
orders_batch1 = orders[['Order_ID', 'Order_Date', 'Customer_ID', 'Product_ID', 'Quantity', 'Payment_Method', 'Order_Status']].head(5)
orders_batch2 = orders[['Order_ID', 'Order_Date', 'Customer_ID', 'Product_ID', 'Quantity', 'Payment_Method', 'Order_Status']].tail(5)

concat_demo = pd.concat([orders_batch1, orders_batch2], ignore_index=True)
concat_demo
```

```python

ecommerce_df = orders.merge(customers, on='Customer_ID', how='left') # same column = customer_id in both datasets
ecommerce_df = ecommerce_df.merge(products, on='Product_ID', how='left')#same column = product_id in both datasets

ecommerce_df['Order_Type'] = ecommerce_df['Quantity'].apply(lambda x: 'Bulk' if x >= 3 else 'Single/Small')
ecommerce_df['Total_Price'] = ecommerce_df.apply(lambda row: row['Quantity'] * row['Unit_Price'], axis=1)


ecommerce_df[['Order_ID', 'Customer_Name', 'Membership_Type', 'Order_Type', 'Total_Price']].head()
```

```python
ecommerce_df['Order_Date'] = pd.to_datetime(ecommerce_df['Order_Date'])

ecommerce_df['Month'] = ecommerce_df['Order_Date'].dt.month
ecommerce_df['Month_Name'] = ecommerce_df['Order_Date'].dt.month_name()
ecommerce_df['Day'] = ecommerce_df['Order_Date'].dt.day
ecommerce_df['Day_of_Week'] = ecommerce_df['Order_Date'].dt.day_name()

ecommerce_df[['Order_ID', 'Order_Date', 'Month_Name', 'Day', 'Day_of_Week']].head()
```

```python
ecommerce_df.to_csv('Ecommerce_Dataset.csv', index=False)
print("Ecommerce dataset has been saved to 'Ecommerce_Dataset.csv' successfully.")
```

## 2. Data Cleaning & Transformations

```python
data["Delay_Minutes"] = data["Delay_Minutes"].fillna(
    data["Delay_Minutes"].median()
)

data["Delay_Minutes"].isnull().sum()
```

```python
data["Flight_Date"] = pd.to_datetime(data["Flight_Date"])
data.head()
```

```python
data = data.drop_duplicates()
data.shape
```

```python
data["Load_Factor"] = (data["Passengers"] / data["Seat_Capacity"]) * 100
data["Month"] = data["Flight_Date"].dt.month
data.head()
```

## 3. Selecting & Filtering Relevant Data

```python
pricy_tickets = data[
    (data["Average_Ticket_Price"] > 6000) &
    (
        data["Origin"].isin(["Bengaluru", "Mumbai", "Srinagar"]) |
        data["Destination"].isin(["Bengaluru", "Mumbai", "Srinagar"])
    ) &
    (data["Travel_Class"] == "Premium Economy")
]
pricy_tickets
```

```python
customer_satisfication = data[data["Passenger_Satisfaction"] >= 4]
customer_satisfication.value_counts()
```

```python
delay_flights = data[data["Delay_Minutes"] > 50]
delay_flights
```

## 4. Sorting Records

```python
data.sort_values(
    by="Average_Ticket_Price",
    ascending=False
)[[
    "Flight_ID",
    "Airline",
    "Origin",
    "Destination",
    "Travel_Class",
    "Average_Ticket_Price"
]].head(5)
```

```python
top_10_highest_delayed_flights = data.sort_values("Delay_Minutes", ascending=False).head(10)
top_10_highest_delayed_flights
```

## 5. Grouping & Aggregation

```python
main_summary = data.groupby(
    ["Origin", "Destination"]
).agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Ticket_Price=("Average_Ticket_Price", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).round(2)

main_summary
```

```python
weather_summary = data.groupby("Weather").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).round(2)

weather_summary
```

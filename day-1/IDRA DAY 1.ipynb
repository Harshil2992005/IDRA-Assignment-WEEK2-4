
## Assignment: Simple Calculator (Day 1)

Create a Python program that accepts two numbers from the user and performs all the basic arithmetic operations on them. Your program should display the results of addition (+), subtraction (-), multiplication (*), division (/), floor division (//), modulus (%), and exponentiation (**). Use appropriate variable names, convert user input to numeric data types, and display the output clearly using formatted print statements.

```python
num1 = int(input("\n enter a first number for an Arithmatic Opeartaion:"))

num2 = int(input("\n enter a second number for an Arithmatic Opeartaion:"))


print("\n sum of to numbers:" , num1 + num2)
print("\n minus of to numbers:" , num1 - num2)
print("\n multiplication of to numbers:" ,num1 * num2)

if num2 == 0:
    print("\n Divison of to numbers cant perform due to second no as zero ")
    print("\n Floor Divison  of to numbers cant perform due to second no as zero ")
    print("\n modulo of to numbers cant perform due to second no as zero ")

else:
    print("\n Divison  of two numbers:" , num1 / num2)
    print("\n Floor Divison  of two numbers:" , num1 // num2)
    print("\n Modulo of two numbers:" , num1 % num2)


print("\n exponentiation of two numbers : " , num1 ** num2)
```

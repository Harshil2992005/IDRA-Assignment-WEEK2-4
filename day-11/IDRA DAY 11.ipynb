```python
import pandas as pd
```

```python
data = pd.read_csv("Day11_Messy_Company_Employee_Dataset.csv")
data.head(10)
```

```python
null_values = data.isnull().sum()
null_values
```

```python
before_cleaning= data.isnull().sum().sum()
print("Missing values before cleaning:",before_cleaning)

before_duplicates_cleaned = data.duplicated().sum()
print("Duplicates before cleaning:",before_duplicates_cleaned)
print(data.duplicated().sum())

before_raws_cleaned = data.shape[0]
print("before shape of raw:",before_raws_cleaned)

print(data.dtypes)
```

```python
data.info()
```

```python
data.describe(include='all')
```

```python
data["Department"] = data["Department"].str.strip().str.title()
data["Department"] = data["Department"].fillna(data["Department"].mode()[0])
data.head()

```

```python

data["Age"] = data["Age"].fillna(data["Age"].median())
data.head()
```

```python
data["Gender"] = data["Gender"].str.strip().str.title()
data["Gender"] = data["Gender"].fillna(data["Gender"].mode()[0])
data.head()
```

```python

data["Annual_Salary"] = data["Annual_Salary"].fillna(data["Annual_Salary"].median())
data.head()
```

```python

data["Experience_Years"] = data["Experience_Years"].fillna(data["Experience_Years"].median())
data.head()
```

```python

data["Performance_Score"] = data["Performance_Score"].fillna(data["Performance_Score"].median())
data.head()
```

```python
data["City"] = data["City"].str.strip().str.title()
data["City"] = data["City"].ffill().bfill()
data.head()
```

```python
data["Work_Mode"] = data["Work_Mode"].str.strip().str.title()
data["Work_Mode"] = data["Work_Mode"].ffill()
data.head()
```

```python
data = data.drop_duplicates()
data
```

```python
data["Joining_Date"] = pd.to_datetime(data["Joining_Date"])
data
```

```python
data.nunique()
```

```python
after_cleaning = data.isnull().sum().sum()
print("Missing values after cleaning:",after_cleaning)

after_duplicates_cleaned = data.duplicated().sum()
print("Duplicates after cleaning:",after_duplicates_cleaned)


after_raws_cleaned = data.shape[0]
print("Final shape of raw:",after_raws_cleaned)

print(data.dtypes)

```

```python

# Comparison
print("BEFORE CLEANING")
print("Rows:", before_raws_cleaned)
print("Missing Values:", before_cleaning)
print("Duplicates:", before_duplicates_cleaned )


print("\nAFTER CLEANING")
print("Rows:", after_raws_cleaned)
print("Missing Values:", after_cleaning)
print("Duplicates:", after_duplicates_cleaned)
```

```python
data.to_csv("cleaned_Day_11.csv")
print("dataset is cleaned and stored in:cleaned_Day_11.csv" )
```

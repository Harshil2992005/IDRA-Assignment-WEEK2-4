```python
sorted_multicol = df.sort_values(by=['Category', 'Total_Sales'], ascending=[True, False])
sorted_multicol
```

```python

highest_sales_product = (df.groupby('Product')['Total_Sales'].sum().sort_values(ascending=False).head(1))
print("Highest Revenue Product:\n", highest_sales_product)
```

```python

highest_sales_category = (df.groupby('Category')['Total_Sales'].sum().sort_values(ascending=False).head(1))
print("Highest Revenue Category:\n", highest_sales_category)
```

```python

cat_stats = df.groupby('Category').agg({'Total_Sales': ['sum', 'mean', 'min', 'max', 'count'],'Quantity': 'sum'})
cat_stats.columns = ['Total_Sales', 'Avg_Sales', 'Min_Sales', 'Max_Sales', 'Num_Orders', 'Total_Qty']
cat_stats
```

```python

city_stats = df.groupby('City').agg({'Total_Sales': ['sum', 'mean', 'count'],'Quantity': 'sum'})
city_stats.columns = ['Total_Sales', 'Avg_Sales', 'Num_Orders', 'Total_Qty']
city_stats.sort_values('Total_Sales', ascending=False)
```

```python

payment_stats = df.groupby('Payment_Method').agg({
    'Total_Sales': ['sum', 'mean', 'count']
})
payment_stats.columns = ['Total_Sales', 'Avg_Sales', 'Num_Orders']
payment_stats.sort_values('Total_Sales', ascending=False)
```

```python

region_stats = df.groupby('Region').agg({'Total_Sales': ['sum', 'mean', 'min', 'max', 'count'],'Quantity': 'sum'})
region_stats.columns = ['Total_Sales', 'Avg_Sales', 'Min_Sales', 'Max_Sales', 'Num_Orders', 'Total_Qty']
region_stats.sort_values('Total_Sales', ascending=False)
```

```python

top_five_products = df.groupby('Product')['Total_Sales'].sum().nlargest(5)
top_five_products
```

```python

top_five_categories = df.groupby('Category')['Total_Sales'].sum().nlargest(5)
top_five_categories
```

```python

qty_per_product = df.groupby('Product')['Quantity'].sum().sort_values(ascending=False)
qty_per_product
```


## Assignment: Number Guessing Game (Day 2)

Create a Python program that generates a random number within a specified range and asks the user to guess it. The program should provide feedback indicating whether the guess is too high, too low, or correct. Limit the player to a fixed number of attempts, and display an appropriate success or failure message at the end of the game. Use conditional statements, loops, and the random module to implement the game logic, while ensuring the program is clear, user friendly, and well organized.


```python
import random 

lower = 1
upper = 50

secret = random.randint(lower, upper)
attempts = 10

for turn in range(1, attempts + 1):
    guess = int(input("\n enter a number form your side!"))
    if guess > secret:
        print("its too high!!")
    elif guess < secret:
        print("its too low!!")
    else:
        print("you guess Correct no !")
        break

if guess == secret:
    print("your winnerr!!!")
else:
    print(f"your attempts limits reached... sorry try again!!: {secret} ")
```
